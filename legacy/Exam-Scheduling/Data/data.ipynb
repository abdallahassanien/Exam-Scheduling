```python
import pandas as pd
```

```python
df = pd.read_excel('data.xlsx')
```

```python
df["num"] = df["Comments"].str.extract(r"(\d+)\s+students").astype(int)
```

```python
df_sum = df.groupby("Subject", as_index=False).agg({
    "num": "sum",
    "Student Sets": lambda x: " ".join(dict.fromkeys(x))
})
```
